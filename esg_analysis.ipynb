###  Fixing Null Consumption by Grouped Median

```python
grouped_median = df.groupby(
    ['facility_id', 'meter_type', 'month']
)['consumption'].transform('median')

df['consumption'] = df['consumption'].fillna(grouped_median)
df['consumption'].isnull().sum()
```

###  Fallback Fill If Any Nulls Remain

```python
meter_median = df.groupby('meter_type')['consumption'].transform('median')
df['consumption'] = df['consumption'].fillna(meter_median)
df['consumption'].isnull().sum()
```

### Removing Duplicates

```python
df = df.drop_duplicates(keep = 'first').reset_index(drop=True)
print(f"Rows after removing dupicates: {len(df):,}")
df.duplicated().sum()

```

### Fixing Region Casing

```python
df['region'] = df['region'].str.strip().str.title()
df['region'].value_counts
```

### Removing null consumption

```python
df = df[df['consumption'] >= 0].reset_index(drop=True)
print(f"Rows after removing negatives: {len(df):,}")
```

### Recalculating emissions_tco2e

```python
emission_factors = {
    'Natural Gas':      2.02  / 1000,
    'Diesel':           2.68  / 1000,
    'Grid Electricity': 0.45  / 1000,
    'Renewable Energy': 0.01  / 1000,
    'Water':            0.0,
    'Waste':            0.0,
} 
df['emissions_tco2e'] = df.apply(
    lambda row : round(
        row['consumption'] * emission_factors.get(row['meter_type'], 0), 4
    ) ,axis = 1
)
df['emissions_tco2e'].describe()
```

### Fixing Data types

```python
df.dtypes
```

```python
df['consumption']        = df['consumption'].astype(float).round(4)
df['emissions_tco2e']    = df['emissions_tco2e'].astype(float).round(4)
df['fiscal_period']      = pd.to_datetime(df['fiscal_period'], format='%Y-%m')
df.dtypes
```

### Add scope_label Column

```python
df['scope_label'] = df['scope'].map({
    'Scope 1' : 'Direct Emissions',
    'Scope 2' : 'Indirect Emissions',
    'N/A'     : 'Non-GHG Metric'
})
df['scope_label'].value_counts()
```

###  Adding is_renewable Column

```python
df['is_renewable'] = (df['meter_type'] == 'Renewable Energy')
df['is_renewable'].value_counts()
```

### Adding emissions_per_employee Column

```python
df['emissions_per_employee'] = (
    df['emissions_tco2e'] / df['facility_employees']
).round(6)

df['emissions_per_employee'].describe()
```

### Checking annual spends 

```python
print(f"Negative annual spend: {(df_sup['annual_spend_usd'] < 0).sum()}")
print(df_sup['annual_spend_usd'].describe())
```

### Initial rows count

```python
intial_rows_sup = len(df_sup)
print(f"Initial Rows: {intial_rows_sup}")
```

### Fixing Null audit_score (Grouped Median)

```python
grouped_median_sup = df_sup.groupby(
    ['risk_profile', 'year']
)['audit_score'].transform('median')

df_sup['audit_score'] = df_sup['audit_score'].fillna(grouped_median_sup)
df_sup['audit_score'].isnull().sum()
```

### Fallback Fill if Nulls remain 

```python
risk_median = df_sup.groupby('risk_profile')['audit_score'].transform('median')
df_sup['audit_score'] = df_sup['audit_score'].fillna(risk_median)
df_sup['audit_score'].isnull().sum()
```

### Removing duplicates

```python
df_sup = df_sup.drop_duplicates(keep='first').reset_index(drop=True)
print(f'Rows after dedup: {len(df_sup):,}')
df_sup.duplicated().sum()
```

### Clipping Score Columns to 0-100

```python
score_cols = [
    'audit_score',
    'labor_compliance_pct',
    'env_compliance_pct',
    'safety_compliance_pct'
]

for col in score_cols:
    df_sup[col] = df_sup[col].clip(0, 100).round(2)

df_sup[score_cols].describe()
```

### Removing Negative annual spend

```python
df_sup = df_sup[df_sup['annual_spend_usd'] >= 0].reset_index(drop=True)
print(f'Rows after spend fix: {len(df_sup):,}')
```

###  Recalculating audit_result From Clean Scores

```python
def get_audit_result(score):
    if score >= 85:
        return 'Approved'
    elif score >= 65:
        return 'Conditional Approval'
    elif score > 50:
        return 'Requires Approval'
    else:
        return 'Failed'
    
df_sup['audit_result'] = df_sup['audit_score'].apply(get_audit_result)
df_sup['audit_result'].value_counts()
```

### Fixing Datatypes

```python
df_sup['year']                 = df_sup['year'].astype(int)
df_sup['audit_score']          = df_sup['audit_score'].astype(float).round(2)
df_sup['labor_compliance_pct'] = df_sup['labor_compliance_pct'].astype(float).round(2)
df_sup['env_compliance_pct']   = df_sup['env_compliance_pct'].astype(float).round(2)
df_sup['safety_compliance_pct']= df_sup['safety_compliance_pct'].astype(float).round(2)
df_sup['annual_spend_usd']     = df_sup['annual_spend_usd'].astype(float).round(2)
df_sup['is_critical']          = df_sup['is_critical'].astype(bool)
df_sup['corrective_action']    = df_sup['corrective_action'].astype(bool)
df_sup.dtypes
```

### Adding compliance_tier Column

```python
df_sup['compliance_tier'] = pd.cut(
    df_sup['audit_score'],
    bins=[0, 50, 65, 85, 100],
    labels=['Failed', 'Requires Improvement', 'Conditional', 'Approved']
)
df_sup['compliance_tier'].value_counts()
```

### Adding spend_tier Column

```python
df_sup['spend_tier'] = pd.cut(
    df_sup['annual_spend_usd'].astype(float),
    bins=[0, 100000, 500000, 2000000, float('inf')],
    labels=['Small', 'Medium', 'Large', 'Strategic'],
    include_lowest=True
)

df_sup['spend_tier'].value_counts(dropna=False)
```

### Adding high_risk_flag_column

```python
df_sup['high_risk_flag'] = (
    (df_sup['risk_profile'] == 'High Risk') &
    (df_sup['audit_score'] < 65)
)

print(f"High risk suppliers flagged: {df_sup['high_risk_flag'].sum()}")
```

### Final Validation

```python
print(f"Initial rows:             {intial_rows_sup:,}")
print(f"Final rows:               {len(df_sup):,}")
print(f"Rows removed:             {intial_rows_sup - len(df_sup):,}")
print(f"Nulls remaining:          {df_sup.isnull().sum().sum()}")
print(f"Duplicates remaining:     {df_sup.duplicated().sum()}")
print(f"Scores outside 0-100:     {((df_sup['audit_score'] < 0) | (df_sup['audit_score'] > 100)).sum()}")
print(f"Unique suppliers:         {df_sup['supplier_id'].nunique()}")
print(f"Unique risk profiles:     {df_sup['risk_profile'].unique()}")
print(f"Unique audit results:     {df_sup['audit_result'].unique()}")
print(f"Years covered:            {sorted(df_sup['year'].unique())}")
print(f"Columns:                  {df_sup.shape[1]}")
```

```python
df_sup['audit_result'] = df_sup['audit_result'].replace(
    'Requires Approval', 'Requires Improvement'
)
print(df_sup['audit_result'].value_counts())
```

```python
df_sup.to_csv('data/esg_supplier_compliance_clean.csv', index=False)
print(f"Exported: {len(df_sup):,} rows × {df_sup.shape[1]} columns")
```
